```python

# This pipeline writes summaries of each table as headers under item's md file

def pipeline(item, force = False):
    #print(item.citekey)
    
    #folder = outputs_root + "/" + item.citekey
    # if os.path.isdir(folder) :
    #     print("skipping existing item", folder, "from",  item.repository.lower() )
    #     return # None, folder
    
    dfs = DataIO.download(item)

    itemname = item.citekey
    document = f"Title: {item.title}\n\n" +\
               f"URL: [{item.url}]({item.url})\n\n" +\
               f"Repository: {item.repository}\n\n"
    imgs = {}
    summs = {}
    for dfname, df in dfs.items():
        dfname = os.path.splitext(dfname)[0]

        docname = os.path.join(outputs_root , itemname + '.md')

        if os.path.exists(docname) and not force:
            print (f"Skipping summarization - already exists: '{docname}'")
            continue

        try:
            summ = Viz.summary(df, hist_fmt='png')
        except Exception as e:
            summ = pd.DataFrame([["oops",b"oobs","Error during summarization", str(e)]], columns=["Column", "Hist", "Error", "Message"])
            warnings.warn(f"Error summarizing {dfname}: {e}")
        summs[dfname] = summ
        

        md, assets = DataIO.to_md(summ, itemname,dfname)
        document += f"## Table: {dfname}\n\n"
        document += md + "\n\n"
        imgs.update(assets)

    DataIO.md_save(outputs_root, itemname, document, imgs)
    print("saved", itemname)

    return dfs, summs
        
```
