```python
def TaxiObservationSpace():
    observation_space = {}

    for taxi_row in range(5):
        for taxi_col in range(5):
            for passenger_location in range(5):
                for destination in range(4):
                    state_num = ((taxi_row * 5 + taxi_col) * 5 + passenger_location) * 4 + destination
                    
                    state = {"State_Num": state_num,
                            "Taxi_Row": taxi_row, 
                            "Taxi_Col": taxi_col, 
                            "Passenger_Location": passenger_location, 
                            "Destination": destination}

                    observation_space[state_num] = state


    states_map = pd.DataFrame(observation_space).T
    return states_map

states_map = TaxiObservationSpace()
states_map
```

```python

actions = "down,up,right,left,pickup,dropoff"

#actions = "v"

def summarize_Q(episode):
    Q_t = pd.DataFrame(episode["internal_state"]).T
    Q_t.columns = actions.split(",")
    Q_t['state'] = Q_t.index.array
    Q_t['episode'] = episode.n
    Q_t['is_training'] = episode.is_training
    return Q_t

def Q_evolution(episodes):
    Qs = [summarize_Q(ep) for i, ep in episodes.iterrows()]

    Qs = pd.concat(Qs)

    eps = pd.DataFrame(range(len(episodes)), columns=["epsd"])

    q = f"""
    SELECT q.episode, Taxi_Row, Taxi_Col
    , {actions}
    , Passenger_Location, Destination
    , is_training
    FROM states_map sm
    CROSS JOIN eps
    LEFT JOIN Qs as q ON sm.State_Num = q.state
    AND eps.epsd = q.episode
    ORDER BY q.episode
    """
    joined = sqldf(q).fillna(0)

    acts = [ f"round(sum({a}),2) as {a}" for a in actions.split(",")]
    acts = "\n,".join(acts)
    q = f"""
    SELECT episode, 
    Taxi_Row, Taxi_Col, Passenger_Location ,
    {acts}
    --, round(sum(up),2) as up -- TODO: Add other directions and preferred one
    --, round(sum(pickup),2) as pickup
    --, count(pickup) as cnt

    FROM joined
    GROUP BY episode, Taxi_Row, Taxi_Col, Passenger_Location 
    ORDER BY episode
    """
    toplot = sqldf(q)
    return toplot


episodes = pd.DataFrame( exper.telemetry.episodes)
toplot = Q_evolution(episodes)
toplot
```
