Some messages got spread into several columns of the dataframe  
This happened due to inherent structure of messages, which contain all the possible characters,  
The presence of " and , chars in the messages makes separating tricky.  

We can solve it easily by concatenating the columns back into one column

```python
def empty_join(lst, sep = ", "):
  return sep.join([x for x in lst if x])
```

```python
spam.columns = "v1,v2,v3,v4,v5".split(",")
spam["msg"] = spam[['v2', 'v3', 'v4', 'v5']].fillna('').agg(empty_join, axis=1)
spam[["v1","msg"]]
```


  ## 4. Compute simple statistics on the data:   


```python

rows = spam.msg.apply(func=lambda n: re.split("[\d\W]+", n.lower()))
words = sum(rows.values.tolist(),[])

wordsPerMess = rows.apply(func = len)

unique, counts = np.unique(words, return_counts=True)
wc = pd.DataFrame({"token":unique, "count": counts})

tokens = wc.sort_values("count",ascending=False)
```

```python
show = rows[:20].apply(lambda r:  " ".join(r))
print("\n".join(show))
```

```python
ans = f"""
number of SMS messages: {spam.shape[0]},  
number of spams: {sum(spam.v1=="spam")},   
total word count: {len(words)},   
averaged number of words per message: {wordsPerMess.mean():.2f},   
5 most frequent words in the file: {tokens.token.head(5).values},   
number of rare words (appear only once in the file): {sum(tokens["count"] == 1)}.  
"""

display(Markdown(ans))
```

### c) Write what is the difference between the results.

We will start by finding the different tokens between the 2 tokenizators

```python
spam['token_diff'] = spam.apply(lambda row: set(row['spacy_tokens']).symmetric_difference(set(row['nltk_tokens'])), axis=1)
```

```python
df = spam[spam.token_diff!=set()]
df.shape
```

Out of 5571 messages, only 1851 (33%) got tokenized differently.  
Let's check out what these differences are.

We will output the tokenisations for these messages into 2 separate text files and visualize their difference using a standard OS visual diff tool

```python
df.spacy_tokens.apply(" ".join).to_csv("spacy_tokens.csv")
```

```python
df.nltk_tokens.apply(" ".join).to_csv("nltk_tokens.csv")
```

![](tokenisation_diff.png)

Upon examining the diff we can conclude that most of the diffences are in the treatment of the punctuation marks.  
Most notably in non-standard use of them due to slang, informal writing and mistakes, characteristic to SMS message communication    


We can investigate this further.  
Let's examine individual tokens that get tokenized differently

```python

spam['spacy_only'] = spam.apply(lambda row: list(set(row['spacy_tokens']).difference(set(row['nltk_tokens' ]))), axis=1)
spam['nltk_only']  = spam.apply(lambda row: list(set(row['nltk_tokens' ]).difference(set(row['spacy_tokens']))), axis=1)
```

```python
so = spam.explode('spacy_only')[['spacy_only']]

q = """
select spacy_only, count(spacy_only) as cnt
from so 
group by spacy_only
order by cnt desc
limit 50
"""

so = ps.sqldf(q).head(50)
```

```python
no = spam.explode('nltk_only')[['nltk_only']]

q = """
select nltk_only, count(nltk_only) as cnt
from no 
group by nltk_only
order by cnt desc
limit 50
"""

no = ps.sqldf(q).head(50)
```

And so we acquire 2 lists of tokens most frequently being tokenized differently between the two libraries:

```python
together = pd.concat([no,so], axis=1)
together
```

Most of them really appear to be caused by punctuation
