Some messages got spread into several columns of the dataframe  
This happened due to inherent structure of messages, which contain all the possible characters,  
The presence of " and , chars in the messages makes separating tricky.  

We can solve it easily by concatenating the columns back into one column

```python
def empty_join(lst, sep = ", "):
  return sep.join([x for x in lst if x])
```

  ## 10. Compare the results of
  NLTK before and after tokenization, lemmatization and stemming

  BArak


Let's break it down to steps:
* First of all, before tokenization each message contained a full sentence. After tokenization - nltk seperated each sentence to words. We can also see that For example the word don't was separated to: "do", "n't".
* Let's use code to see the differences between tokenization and lemmatization:

```python
# Compare 'tokens_nltk' and 'lemmatized_tokens_nltk' columns and print unique differences
unique_differences = set()  # Use a set to store unique differences

for index, row in spam.iterrows():
    original_tokens = row['tokens_nltk']
    lemmatized_tokens = row['lemmatized_tokens_nltk']

    # Check if there are any differences between the two lists
    if original_tokens != lemmatized_tokens:
        for original_token, lemmatized_token in zip(original_tokens, lemmatized_tokens):
            if original_token != lemmatized_token and (original_token, lemmatized_token) not in unique_differences:
                unique_differences.add((original_token, lemmatized_token))
                print(f"Token '{original_token}' was lemmatized as '{lemmatized_token}'.")
```

We can see that plural suffixes were constantly removed in the lemmatization process. However, they were sometimes unnecessarily removed (For example: the word "was" was turned into "wa" and it should have stayed "was". Now let's compare lemmatization and stemming:

```python
# Compare 'lemmatized_tokens_nltk' and 'stemmed_tokens_nltk' columns and print unique differences
unique_differences_lemmatization_stemming = set()  # Use a set to store unique differences

for index, row in spam.iterrows():
    lemmatized_tokens = row['lemmatized_tokens_nltk']
    stemmed_tokens = row['stemmed_tokens_nltk']

    # Check if there are any differences between the two lists
    if lemmatized_tokens != stemmed_tokens:
        for lemmatized_token, stemmed_token in zip(lemmatized_tokens, stemmed_tokens):
            if lemmatized_token != stemmed_token and (lemmatized_token, stemmed_token) not in unique_differences_lemmatization_stemming:
                unique_differences_lemmatization_stemming.add((lemmatized_token, stemmed_token))
                print(f"Lemmatized token '{lemmatized_token}' was stemmed as '{stemmed_token}'.")
```

We notice the following differences:
* All words were lowercased.
* All suffixes (such as plural suffixes, ing, able, etc.) were removed.
* 'y' suffix was turned into 'i'.

```python
## Step 10 - Compare Tokenization, Lemmatization, and Stemming (Chat Data)

# Create dictionaries to store comparisons between steps
differences_tokenization = set()  # Differences between original text and tokenized
differences_lemmatization = set()  # Differences between tokenized and lemmatized
differences_stemming = set()  # Differences between lemmatized and stemmed

# Iterating through the rows of the chat dataset
for index, row in chat.iterrows():
    original_tokens = row['tokens_nltk']  # Tokenized message
    lemmatized_tokens = row['lemmatized_tokens_nltk']  # Lemmatized message
    stemmed_tokens = row['stemmed_tokens_nltk']  # Stemmed message

    # Compare tokenization with original tokens
    for token in original_tokens:
        if token not in row.iloc[0]:
            differences_tokenization.add(token)

    # Compare lemmatization with tokenized tokens
    for original_token, lemmatized_token in zip(original_tokens, lemmatized_tokens):
        if original_token != lemmatized_token:
            differences_lemmatization.add((original_token, lemmatized_token))

    # Compare stemming with lemmatized tokens
    for lemmatized_token, stemmed_token in zip(lemmatized_tokens, stemmed_tokens):
        if lemmatized_token != stemmed_token:
            differences_stemming.add((lemmatized_token, stemmed_token))

# Display unique differences for each step
print("Differences between Original and Tokenized (NLTK):")
for token in differences_tokenization:
    print(f"Original Token: {token}")

print("\nDifferences between Tokenized and Lemmatized (NLTK):")
for original_token, lemmatized_token in differences_lemmatization:
    print(f"Token '{original_token}' was lemmatized as '{lemmatized_token}'.")

print("\nDifferences between Lemmatized and Stemmed (NLTK):")
for lemmatized_token, stemmed_token in differences_stemming:
    print(f"Lemmatized token '{lemmatized_token}' was stemmed as '{stemmed_token}'.")
```
